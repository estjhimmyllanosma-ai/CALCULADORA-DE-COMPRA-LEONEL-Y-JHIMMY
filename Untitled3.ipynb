```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Lista para guardar los productos
productos = []

titulo = widgets.HTML("<h2>CALCULADORA DE COMPRA</h2>")

txt_producto = widgets.Text(
    description="Producto:",
    placeholder="Escriba el nombre"
)

txt_precio = widgets.FloatText(
    description="Precio (Bs):",
    value=0.0
)

txt_cantidad = widgets.IntText(
    description="Cantidad:",
    value=1
)

btn_registrar = widgets.Button(
    description="Agregar producto",
    button_style="success"
)

salida = widgets.Output()

def registrar_producto(b):
    producto = txt_producto.value.strip()
    precio = txt_precio.value
    cantidad = txt_cantidad.value

    if producto == "":
        with salida:
            clear_output(wait=True)
            print("Escriba el nombre del producto.")
        return

    if precio < 0 or cantidad <= 0:
        with salida:
            clear_output(wait=True)
            print("Revise el precio y la cantidad.")
        return

    total = precio * cantidad

    productos.append({
        "nombre": producto,
        "precio": precio,
        "cantidad": cantidad,
        "total": total
    })

    with salida:
        clear_output(wait=True)

        total_compra = 0

        print("RESUMEN DE LA COMPRA")
        print("------------------------------")

        for i, p in enumerate(productos, start=1):
            print(f"Producto {i}: {p['nombre']}")
            print(f"Precio unitario: Bs {p['precio']:.2f}")
            print(f"Cantidad: {p['cantidad']}")
            print(f"Subtotal: Bs {p['total']:.2f}")
            print("------------------------------")

            total_compra += p["total"]

        print(f"TOTAL DE LA COMPRA: Bs {total_compra:.2f}")

    # Limpiar los campos para ingresar otro producto
    txt_producto.value = ""
    txt_precio.value = 0.0
    txt_cantidad.value = 1

btn_registrar.on_click(registrar_producto)

display(
    titulo,
    txt_producto,
    txt_precio,
    txt_cantidad,
    btn_registrar,
    salida
)
```
